```python
import numpy as np
import random
import matplotlib.pyplot as plt


# Quadtree implementation
class Quadtree:
    def __init__(self, x_min, x_max, y_min, y_max, max_points=4, depth=0):
        self.x_min, self.x_max = x_min, x_max
        self.y_min, self.y_max = y_min, y_max
        self.max_points = max_points
        self.depth = depth
        self.points = []
        self.children = []

    def insert(self, point, index):
        if not (self.x_min <= point[0] <= self.x_max and self.y_min <= point[1] <= self.y_max):
            return False
        
        if len(self.points) < self.max_points and not self.children:
            self.points.append((point, index))
            return True
        
        if not self.children:
            self.subdivide()
        
        for child in self.children:
            if child.insert(point, index):
                return True
        return False
    
    def subdivide(self):
        mid_x = (self.x_min + self.x_max) / 2
        mid_y = (self.y_min + self.y_max) / 2
        self.children = [
            Quadtree(self.x_min, mid_x, self.y_min, mid_y, self.max_points, self.depth + 1),
            Quadtree(mid_x, self.x_max, self.y_min, mid_y, self.max_points, self.depth + 1),
            Quadtree(self.x_min, mid_x, mid_y, self.y_max, self.max_points, self.depth + 1),
            Quadtree(mid_x, self.x_max, mid_y, self.y_max, self.max_points, self.depth + 1),
        ]
        for point, index in self.points:
            for child in self.children:
                if child.insert(point, index):
                    break
        self.points = []
    
    def query_range(self, x, y, radius):
        results = []
        if x + radius < self.x_min or x - radius > self.x_max or y + radius < self.y_min or y - radius > self.y_max:
            return results
        
        for point, index in self.points:
            if np.linalg.norm(np.array(point) - np.array([x, y])) <= radius:
                results.append(index)
        
        if self.children:
            for child in self.children:
                results.extend(child.query_range(x, y, radius))
        return results

class LocalizedStates:
    def __init__(self, N, box_size):
        self.N = N  # Number of states
        self.box_size = box_size  # Size of simulation domain
        self.positions = np.random.rand(N, 2) * box_size  # Random 2D positions
        self.energies = np.random.normal(loc=0, scale=0.1, size=N)  # Gaussian energy distribution in eV
        
        # Build Quadtree
        self.tree = Quadtree(0, box_size, 0, box_size)
        for i in range(N):
            self.tree.insert(self.positions[i], i)
    
    def plot_states(self):
        plt.scatter(self.positions[:, 0], self.positions[:, 1], c=self.energies, cmap='coolwarm')
        plt.colorbar(label="Energy (eV)")
        plt.xlabel("X Position (nm)")
        plt.ylabel("Y Position (nm)")
        plt.title("Localized States in AlGaAs")
        plt.show()

def miller_abrahams_rate(E_i, E_j, r_ij, a, nu_0, kT):
    if E_j > E_i:
        return nu_0 * np.exp(-2 * r_ij / a) * np.exp(-(E_j - E_i) / kT)
    else:
        return nu_0 * np.exp(-2 * r_ij / a)

class KMC_Simulator:
    def __init__(self, states, a=5.0, nu_0=1e13, kT=0.025):
        self.states = states  # Localized states object
        self.a = a  # Localization length (nm)
        self.nu_0 = nu_0  # Attempt frequency (Hz)
        self.kT = kT  # Thermal energy (eV)
        self.time = 0  # Simulation time
        self.history = []  # Store event history
    
    def run_step(self):
        idx = random.randint(0, self.states.N - 1)  # Pick a random state
        pos_i, E_i = self.states.positions[idx], self.states.energies[idx]
        
        neighbors_idx = self.states.tree.query_range(pos_i[0], pos_i[1], 10)  # Find nearby states in radius 10 nm
        if len(neighbors_idx) < 2:
            return  # Skip if no valid neighbor
        
        rates = []
        for j in neighbors_idx:
            if j != idx:
                r_ij = np.linalg.norm(self.states.positions[j] - pos_i)
                rate = miller_abrahams_rate(E_i, self.states.energies[j], r_ij, self.a, self.nu_0, self.kT)
                rates.append((j, rate))
        
        if not rates:
            return  # Skip if no transitions
        
        total_rate = sum(r for _, r in rates)
        if total_rate == 0:
            return
        
        r_select = random.uniform(0, total_rate)
        cumulative = 0
        for j, rate in rates:
            cumulative += rate
            if cumulative >= r_select:
                self.states.positions[idx] = self.states.positions[j]  # Move carrier
                self.states.energies[idx] = self.states.energies[j]  # Update energy
                break
        
        dt = -np.log(random.random()) / total_rate
        self.time += dt
        self.history.append((self.time, idx))
    
    def run_simulation(self, steps=1000):
        for _ in range(steps):
            self.run_step()
    
    def plot_trajectory(self):
        times, indices = zip(*self.history)
        plt.plot(times, indices, marker="o", linestyle="None", alpha=0.5)
        plt.xlabel("Time (s)")
        plt.ylabel("State Index")
        plt.title("Carrier Hopping Events Over Time")
        plt.show()

states = LocalizedStates(N=500, box_size=100)
states.plot_states()

sim = KMC_Simulator(states, a=5.0, nu_0=1e13, kT=0.025)
sim.run_simulation(steps=1000)
sim.plot_trajectory()
```

```python
import numpy as np  
import random  
import matplotlib.pyplot as plt  
from PIL import Image  
import os  

# Quadtree:
class Quadtree:
    def __init__(self, x_min, x_max, y_min, y_max, max_points=4, depth=0):
        self.x_min, self.x_max = x_min, x_max  # x ašies riba
        self.y_min, self.y_max = y_min, y_max  # y ašies riba
        self.max_points = max_points  # Maksimalus dalelių skaičius viename node
        self.depth = depth  # Nusako quadtree gylį (nusako nodų skaičių)
        self.points = []  # taškai viename node
        self.children = []  # child nodai

    def insert(self, point, index):
        # Patikrina ar taškas tam tikro nodo ribose
        if not (self.x_min <= point[0] <= self.x_max and self.y_min <= point[1] <= self.y_max):
            return False  
        
        # Jei yra vietos ir nėra child nodo jis sukūria tašką laisvoje vietoje
        if len(self.points) < self.max_points and not self.children:
            self.points.append((point, index))  
            return True
        
        #Jei nėra child nodo jis padalija nodą į 4 child nodus
        if not self.children:
            self.subdivide()
        
        # įkelia tašką į vieną iš child nodų
        for child in self.children:
            if child.insert(point, index):
                return True
        return False
    
    def subdivide(self):
        # Padalina nodą į 4 child nodus ir suranda jų vidurio tašką
        mid_x = (self.x_min + self.x_max) / 2
        mid_y = (self.y_min + self.y_max) / 2
        self.children = [
            Quadtree(self.x_min, mid_x, self.y_min, mid_y, self.max_points, self.depth + 1),
            Quadtree(mid_x, self.x_max, self.y_min, mid_y, self.max_points, self.depth + 1),
            Quadtree(self.x_min, mid_x, mid_y, self.y_max, self.max_points, self.depth + 1),
            Quadtree(mid_x, self.x_max, mid_y, self.y_max, self.max_points, self.depth + 1),
        ]  # Sukuria 4 child nodus
        for point, index in self.points:  # Perkelia taškus iš parent nodo į child nodą
            for child in self.children:
                if child.insert(point, index):
                    break
        self.points = []  # Ištrina taškus iš parent nodo.
    
    def query_range(self, x, y, radius): #query_range ieško tam tikrų taškų Quadtree erdvėje
        results = []  
        # Patikrina ar tai įeina į Quadtree ribas 
        if x + radius < self.x_min or x - radius > self.x_max or y + radius < self.y_min or y - radius > self.y_max:
            return results  
        
        # Patikrina taškus mazge ar jie įeina į spindulį
        for point, index in self.points:
            if np.linalg.norm(np.array(point) - np.array([x, y])) <= radius:
                results.append(index)
        
       
        if self.children:
            for child in self.children:
                results.extend(child.query_range(x, y, radius))  # Čia kaip suprantu Quadtree tikrina kurie "children" turi tam tikram spinduly dalelių kurie ne tuos atmeta.
        return results


# Klasė kuri sukūria daleles ir parenka jiems lokalizuotas būsens, energiją ir poziciją.
class LocalizedStates:
    def __init__(self, N, box_size):
        self.N = N  # Lokalizuotų būsenų skaičius (dalelių skaičius)
        self.box_size = box_size  # Dydis tos erdvės kur yra dalelės
        
        # Parenka tiek pozicijų kiek yra dalelių
        self.positions = np.random.rand(N, 2) * box_size  # Parenka pozicijas
        # Sukūria energijas nuo -0.1 iki 0
        self.energies = np.random.uniform(-0.1, 0.0, size=N)
        
        # Sukūria Quadtree pozicijas
        self.tree = Quadtree(0, box_size, 0, box_size)
        for i in range(N):
            self.tree.insert(self.positions[i], i)  # Įrašo kiekvieną dalelę į Quadtree
    
    def update_plot(self, ax, colorbar=False):  #plotinimas
        ax.clear()  # Clear the axes
        scatter = ax.scatter(self.positions[:, 0], self.positions[:, 1], c=self.energies, cmap='coolwarm', vmin=-0.1, vmax=0.0)  # Plot positions with energies
        ax.set_xlabel("X Position (nm)")  
        ax.set_ylabel("Y Position (nm)")  
        ax.set_title("Localized States in AlGaAs")  
        
        if colorbar:   #Colorbaras
            cbar = ax.figure.colorbar(scatter, ax=ax, label="Energy (eV)")
        
        return scatter


# KMC klasė
class KMC_Simulator:
    def __init__(self, states, a=5.0, nu_0=1e13, kT=0.025):
        self.states = states  # Lokalizacijos būsenos
        self.a = a  # Lokalizacijos ilgis (nm)
        self.nu_0 = nu_0  # Dažnis (Hz)
        self.kT = kT  # Energija (eV)
        self.time = 0  # Simuliacijos laikas
    
    def run_step(self):
        idx = random.randint(0, self.states.N - 1)  # Paima bet kurią lokalizuotą būsena
        pos_i, E_i = self.states.positions[idx], self.states.energies[idx]  #Išsaugo jos poziciją ir energiją
        
        #Pagal 10nm atstumą ieško būsenų į kurias gali peršokti tame Quadtree
        neighbors_idx = self.states.tree.query_range(pos_i[0], pos_i[1], 10)
        if len(neighbors_idx) < 2:  #Jei nėra galimų būsenų tada praleidžia
            return
        
        rates = []  # Čia išsaugo šokinėjimus, kad paskui gauti jų dažnį
        for j in neighbors_idx:  # Pereina per kaimynines būsenas kad patikrintų ar gali peršokti
            if j != idx:
                r_ij = np.linalg.norm(self.states.positions[j] - pos_i)  # Suskaičiuoja atstumą tarp pozicijų
                rate = miller_abrahams_rate(E_i, self.states.energies[j], r_ij, self.a, self.nu_0, self.kT)  # Pagal Millerio-Abramso formulę yra suskaičiuojamas Millerio-Abramso dažnis
                rates.append((j, rate))
        
        if not rates:  # Jei nėra jokių šokinėjimų tada tiesiog praleidžia tai
            return
        
        total_rate = sum(r for _, r in rates)  # suskaičiuojamas šokinėjimo dažnis
        if total_rate == 0:
            return
        
        r_select = random.uniform(0, total_rate)  # Tam tikra vertė kuri nusako šokinėjimo dažnį
        cumulative = 0
        for j, rate in rates:
            cumulative += rate
            if cumulative >= r_select:  # Kai cumulative tampa didesnis už tam tikrą vertę tada dalelė peršoka į kitą būseną
                self.states.positions[idx] = np.random.rand(2) * self.states.box_size  # Monte-Carlo parenka nauja poziciją
                self.states.energies[idx] = self.states.energies[j]  # Po persikelimo gauna tą pačią energiją kaip ir dalelė kuri buvo čia prieš tai
                break
        
        dt = -np.log(random.random()) / total_rate  # delta t suskaičiojamas pagal total rate
        self.time += dt
    
    def run_simulation(self, steps=100, save_path="exciton_simulation.gif"): #Simuliacijos funkcija
        fig, ax = plt.subplots(figsize=(6, 6))  
        
        frames = []  # Plotų(framų) listas
        
        self.states.update_plot(ax, colorbar=True)  #Pradinis plotas nusako koksai bus colorbar, kad nesikeistų per daug
        
        for _ in range(steps):  # Simuliacija veikia tiek kiek jai yra duota žingsnių
            self.run_step()  #KMC Simuliacija veikia
            scatter = self.states.update_plot(ax, colorbar=False)  #Atnaujina plotą
            
            frame_path = f"frame_{_}.png"  # Išsaugo kiekvieną Frame'ą gifo kūrimui
            plt.savefig(frame_path)
            frames.append(frame_path)
        
        self.create_gif(frames, save_path)  #Sukuria gifą iš framų
        
        for frame in frames:  
            os.remove(frame)
    
    def create_gif(self, frames, save_path): #Gifo kurimo funkcija
        images = []  
        for frame in frames:
            img = Image.open(frame)  
            images.append(img)
        
        images[0].save(save_path, save_all=True, append_images=images[1:], duration=100, loop=0)  
        print(f"GIF saved as {save_path}")  


def miller_abrahams_rate(E_i, E_j, r_ij, a, nu_0, kT): #Miller abramso formulė kuri nusako galimybę(šansą) dalalei peršokti
    if E_j > E_i:
        return nu_0 * np.exp(-2 * r_ij / a) * np.exp(-(E_j - E_i) / kT)  
    else:
        return nu_0 * np.exp(-2 * r_ij / a)  

states = LocalizedStates(N=100, box_size=100) #Kiek dalelių ir kiek lokalizuotų būsenų skaičius (100x100)
sim = KMC_Simulator(states, a=5.0, nu_0=1e13, kT=0.00001) #KMC simulatorius simuliuoja jų šokinėjimą pagal nu, kT ir a parametrus
sim.run_simulation(steps=100, save_path="exciton_simulation.gif") #sukuria 100 framu iš kurių padaro simulacijos gifą
```

```python
import numpy as np  # Import numpy for numerical calculations
import random  # Import random library for generating random numbers
import matplotlib.pyplot as plt  # Import matplotlib for plotting
from PIL import Image  # Import Pillow for working with images
import os  # Import os library for interacting with the file system

# Quadtree implementation for spatial indexing
class Quadtree:
    def __init__(self, x_min, x_max, y_min, y_max, max_points=4, depth=0):
        self.x_min, self.x_max = x_min, x_max  # Set the x-axis boundaries
        self.y_min, self.y_max = y_min, y_max  # Set the y-axis boundaries
        self.max_points = max_points  # Maximum number of points allowed in a node
        self.depth = depth  # Depth of the quadtree
        self.points = []  # List to store points in the node
        self.children = []  # List to store child nodes

    def insert(self, point, index):
        # Check if the point is within the boundaries of the current node
        if not (self.x_min <= point[0] <= self.x_max and self.y_min <= point[1] <= self.y_max):
            return False  # Return False if the point is out of bounds
        
        # If there is space in the current node and no children, insert the point here
        if len(self.points) < self.max_points and not self.children:
            self.points.append((point, index))  # Append the point and its index
            return True
        
        # If there are no children, subdivide the node into four children
        if not self.children:
            self.subdivide()
        
        # Try to insert the point into one of the children
        for child in self.children:
            if child.insert(point, index):
                return True
        return False
    
    def subdivide(self):
        # Split the node into four children by finding the midpoint of the node
        mid_x = (self.x_min + self.x_max) / 2
        mid_y = (self.y_min + self.y_max) / 2
        self.children = [
            Quadtree(self.x_min, mid_x, self.y_min, mid_y, self.max_points, self.depth + 1),
            Quadtree(mid_x, self.x_max, self.y_min, mid_y, self.max_points, self.depth + 1),
            Quadtree(self.x_min, mid_x, mid_y, self.y_max, self.max_points, self.depth + 1),
            Quadtree(mid_x, self.x_max, mid_y, self.y_max, self.max_points, self.depth + 1),
        ]  # Create four child nodes
        for point, index in self.points:  # Move points from the parent to the appropriate child
            for child in self.children:
                if child.insert(point, index):
                    break
        self.points = []  # Clear the points from the parent node
    
    def query_range(self, x, y, radius):
        results = []  # List to store the results of the query
        # If the query range is outside the boundaries of the node, return an empty list
        if x + radius < self.x_min or x - radius > self.x_max or y + radius < self.y_min or y - radius > self.y_max:
            return results  # Return empty list if out of bounds
        
        # Check the points in the current node and add them to the results if they are within the query range
        for point, index in self.points:
            if np.linalg.norm(np.array(point) - np.array([x, y])) <= radius:
                results.append(index)
        
        # Recursively check the child nodes for points within the range
        if self.children:
            for child in self.children:
                results.extend(child.query_range(x, y, radius))  # Recursively check child nodes
        return results


# Class for creating localized states and handling their positions and energies
class LocalizedStates:
    def __init__(self, N, box_size):
        self.N = N  # Number of localized states
        self.box_size = box_size  # Size of the simulation domain
        
        # Generate N random positions within the box size
        self.positions = np.random.rand(N, 2) * box_size  # Random positions within the box size
        # Randomly generate energies for each state between -0.1 and 0.0 eV
        self.energies = np.random.uniform(-0.1, 0.0, size=N)
        
        # Create the quadtree for spatial indexing
        self.tree = Quadtree(0, box_size, 0, box_size)
        for i in range(N):
            self.tree.insert(self.positions[i], i)  # Insert each state into the quadtree
    
    def update_plot(self, ax, colorbar=False):
        ax.clear()  # Clear the axes
        # Plot the positions of the localized states with a color map for their energies
        scatter = ax.scatter(self.positions[:, 0], self.positions[:, 1], c=self.energies, cmap='coolwarm', vmin=-0.1, vmax=0.0)
        ax.set_xlabel("X Position (nm)")  # X-axis label
        ax.set_ylabel("Y Position (nm)")  # Y-axis label
        ax.set_title("Localized States in AlGaAs")  # Title of the plot
        
        # Add a colorbar if specified
        if colorbar:
            cbar = ax.figure.colorbar(scatter, ax=ax, label="Energy (eV)")
        
        return scatter


# Kinetic Monte Carlo (KMC) simulation class
class KMC_Simulator:
    def __init__(self, states, a=5.0, nu_0=1e13, kT=0.025):
        self.states = states  # Localized states object
        self.a = a  # Localization length (nm)
        self.nu_0 = nu_0  # Attempt frequency (Hz)
        self.kT = kT  # Thermal energy (eV)
        self.time = 0  # Simulation time
    
    def run_step(self):
        idx = random.randint(0, self.states.N - 1)  # Pick a random localized state
        pos_i, E_i = self.states.positions[idx], self.states.energies[idx]
        
        # Query for nearby states in the quadtree within a radius of 10 nm
        neighbors_idx = self.states.tree.query_range(pos_i[0], pos_i[1], 10)
        if len(neighbors_idx) < 2:  # Skip if no valid neighbors
            return
        
        rates = []  # List to store the transition rates
        for j in neighbors_idx:  # Iterate through neighboring states
            if j != idx:
                r_ij = np.linalg.norm(self.states.positions[j] - pos_i)  # Calculate distance between states
                # Calculate the Miller-Abrahams rate for the transition
                rate = miller_abrahams_rate(E_i, self.states.energies[j], r_ij, self.a, self.nu_0, self.kT)
                rates.append((j, rate))
        
        if not rates:  # If no transitions, skip the step
            return
        
        total_rate = sum(r for _, r in rates)  # Calculate the total rate of transitions
        if total_rate == 0:
            return
        
        # If kT is close to zero, make it even harder for transitions unless energy differences are very small
        if self.kT < 0.0001:
            # If kT is small, the transition rate is extremely low. Force the transition to happen only if energy difference is almost zero.
            for j, rate in rates:
                if np.abs(self.states.energies[j] - E_i) < 0.01:  # Allow transition only if energy difference is small
                    r_select = random.uniform(0, total_rate)
                    cumulative = 0
                    for j, rate in rates:
                        cumulative += rate
                        if cumulative >= r_select:
                            # Update the exciton position and energy
                            self.states.positions[idx] = np.random.rand(2) * self.states.box_size
                            self.states.energies[idx] = self.states.energies[j]
                            break
                    break
            return
        
        # Choose a random transition based on the rates
        r_select = random.uniform(0, total_rate)
        cumulative = 0
        for j, rate in rates:
            cumulative += rate
            if cumulative >= r_select:  # If the cumulative rate exceeds the selected value
                # Update the position (move the carrier to a new position that is not occupied)
                self.states.positions[idx] = np.random.rand(2) * self.states.box_size  # Random new position
                self.states.energies[idx] = self.states.energies[j]  # Transfer energy
                break
        
        # Calculate the time step based on the total rate
        dt = -np.log(random.random()) / total_rate
        self.time += dt
    
    def run_simulation(self, steps=100, save_path="exciton_simulation.gif"):
        fig, ax = plt.subplots(figsize=(6, 6))  # Create a figure for plotting
        
        frames = []  # List to store the image frames for the GIF
        
        # First frame to create the color bar
        self.states.update_plot(ax, colorbar=True)
        
        for _ in range(steps):  # Run the simulation for a specified number of steps
            self.run_step()  # Perform one Monte Carlo step
            scatter = self.states.update_plot(ax, colorbar=False)  # Update the plot
            
            # Save each frame as an image
            frame_path = f"frame_{_}.png"
            plt.savefig(frame_path)
            frames.append(frame_path)
        
        # Create a GIF from the frames
        self.create_gif(frames, save_path)
        
        # Clean up the images after creating the GIF
        for frame in frames:
            os.remove(frame)
    
    def create_gif(self, frames, save_path):
        images = []  # List to store the images
        for frame in frames:
            img = Image.open(frame)  # Open each image
            images.append(img)
        
        images[0].save(save_path, save_all=True, append_images=images[1:], duration=100, loop=0)  # Save the GIF
        print(f"GIF saved as {save_path}")  # Print a message that the GIF has been saved


def miller_abrahams_rate(E_i, E_j, r_ij, a, nu_0, kT):
    # Calculate the transition rate based on the Miller-Abrahams formula
    if E_j > E_i:
        return nu_0 * np.exp(-2 * r_ij / a) * np.exp(-(E_j - E_i) / kT)
    else:
        return nu_0 * np.exp(-2 * r_ij / a)

# Run simulation with a reduced number of steps and only 3 states
states = LocalizedStates(N=100, box_size=100)  # Create 100 localized states
sim = KMC_Simulator(states, a=5.0, nu_0=1e13, kT=0.000001)  # Very low kT value
sim.run_simulation(steps=100, save_path="exciton_simulation.gif")  # Run the simulation for 100 steps and save as GIF
```

```python
import numpy as np
import random
import matplotlib.pyplot as plt

# Quadtree Node definition
class QuadtreeNode:
    def __init__(self, x_min, x_max, y_min, y_max, level=0):
        self.x_min = x_min
        self.x_max = x_max
        self.y_min = y_min
        self.y_max = y_max
        self.level = level
        self.points = []
        self.children = []

    def is_leaf(self):
        return len(self.children) == 0

# Quadtree implementation
class Quadtree:
    def __init__(self, x_min, x_max, y_min, y_max, max_level=5, max_points_per_cell=10):
        self.root = QuadtreeNode(x_min, x_max, y_min, y_max)
        self.max_level = max_level
        self.max_points_per_cell = max_points_per_cell

    def insert(self, point):
        self._insert(self.root, point)

    def _insert(self, node, point):
        # If the node is a leaf and has space, add the point
        if node.is_leaf():
            if len(node.points) < self.max_points_per_cell or node.level == self.max_level:
                node.points.append(point)
                return
            else:
                # Subdivide and distribute points to children
                self.subdivide(node)
                self._insert(node, point)
        else:
            # Insert into the correct child
            for child in node.children:
                if child.x_min <= point[0] < child.x_max and child.y_min <= point[1] < child.y_max:
                    self._insert(child, point)
                    break

    def subdivide(self, node):
        mid_x = (node.x_min + node.x_max) / 2
        mid_y = (node.y_min + node.y_max) / 2
        node.children = [
            QuadtreeNode(node.x_min, mid_x, node.y_min, mid_y, node.level + 1),
            QuadtreeNode(mid_x, node.x_max, node.y_min, mid_y, node.level + 1),
            QuadtreeNode(node.x_min, mid_x, mid_y, node.y_max, node.level + 1),
            QuadtreeNode(mid_x, node.x_max, mid_y, node.y_max, node.level + 1),
        ]
        # Distribute points to children
        for point in node.points:
            for child in node.children:
                if child.x_min <= point[0] < child.x_max and child.y_min <= point[1] < child.y_max:
                    child.points.append(point)
                    break
        node.points.clear()

    def query_nearby(self, x, y, radius):
        """Find all points within a radius around (x, y)."""
        return self._query_nearby(self.root, x, y, radius)

    def _query_nearby(self, node, x, y, radius):
        nearby_points = []
        # Check if the query point is inside this node's bounds
        if not (x + radius < node.x_min or x - radius >= node.x_max or y + radius < node.y_min or y - radius >= node.y_max):
            # Add points from this node if within the radius
            for point in node.points:
                if (x - point[0])**2 + (y - point[1])**2 <= radius**2:
                    nearby_points.append(point)
            # Recursively query child nodes
            if not node.is_leaf():
                for child in node.children:
                    nearby_points.extend(self._query_nearby(child, x, y, radius))
        return nearby_points
```

```python
# Exponential DOS function
def exponential_dos(energy, N, e0):
    return N / e0 * np.exp(-energy / e0)

# Hopping rate function (Miller-Abrahams model)
def hopping_rate(energy_i, energy_j, r_ij, a, n0, T):
    """
    Calculate the hopping rate using the Miller-Abrahams expression.
    :param energy_i: Energy of the initial state
    :param energy_j: Energy of the target state
    :param r_ij: Distance between the states
    :param a: Decay length
    :param n0: Attempt-to-escape frequency
    :param T: Temperature (for the Boltzmann factor)
    :return: Hopping rate between states i and j
    """
    beta = 1 / (T * 1.38e-23)  # Boltzmann factor
    energy_diff = energy_j - energy_i
    rate = n0 * np.exp(-r_ij / a) * np.exp((energy_diff) / (T * 1.38e-23))
    return rate

# Function to create the 100x100 lattice and insert the points in the quadtree
def create_lattice_with_quadtree(N0, N, e0, a, n0, T, grid_size=100):
    lattice = []
    quadtree = Quadtree(0, grid_size, 0, grid_size)
    for x in range(grid_size):
        for y in range(grid_size):
            energy = np.random.exponential(scale=e0)
            lattice.append((x, y, energy))
            quadtree.insert((x, y, energy))
    return lattice, quadtree

# Monte Carlo simulation for exciton hopping and energy relaxation
def monte_carlo_simulation_with_quadtree(lattice, quadtree, grid_size, n0, a, T, num_excitons=10000, radius=1):
    exciton_energies = []
    
    for _ in range(num_excitons):
        # Randomly choose a starting site for the exciton
        initial_x, initial_y, initial_energy = random.choice(lattice)
        exciton_energy = initial_energy

        # Simulate hopping until the exciton recombines (for a simple model)
        while True:
            # Query the quadtree for nearby states
            nearby_states = quadtree.query_nearby(initial_x, initial_y, radius)
            # Filter out the starting point itself
            nearby_states = [state for state in nearby_states if state != (initial_x, initial_y, initial_energy)]
            
            if not nearby_states:
                break

            # Randomly choose a neighboring state
            neighbor = random.choice(nearby_states)
            neighbor_x, neighbor_y, neighbor_energy = neighbor

            # Calculate the distance between the initial site and the chosen neighboring site
            r_ij = np.sqrt((neighbor_x - initial_x)**2 + (neighbor_y - initial_y)**2)
            
            # Calculate the hopping rate between the sites
            rate = hopping_rate(exciton_energy, neighbor_energy, r_ij, a, n0, T)
            
            # Decide if the exciton hops based on the rate (Monte Carlo criterion)
            if random.random() < rate:
                exciton_energy = neighbor_energy
                initial_x, initial_y, initial_energy = neighbor
            
            # Check for recombination (simplified to random decision)
            if random.random() < 0.01:  # Recombination probability
                break
        
        # Record the energy of the exciton at recombination
        exciton_energies.append(exciton_energy)
    
    return np.array(exciton_energies)

# Plot the resulting PL spectrum (exciton recombination energies)
def plot_pl_spectrum(exciton_energies):
    plt.hist(exciton_energies, bins=50, alpha=0.7, color="blue")
    plt.title("Exciton PL Spectrum")
    plt.xlabel("Energy (eV)")
    plt.ylabel("Count")
    plt.show()

# Main simulation function
def run_simulation_with_quadtree():
    # Set simulation parameters
    N0 = 2500  # Number of sites
    N = 100    # Concentration of localized states
    e0 = 0.1   # Energy scale for exponential DOS (eV)
    a = 0.5    # Decay length (nm)
    n0 = 1e13  # Attempt-to-escape frequency (s^-1)
    T = 300    # Temperature (K)
    num_excitons = 10000  # Number of excitons to simulate
    grid_size = 100  # 100x100 grid

    # Create a random lattice and quadtree
    lattice, quadtree = create_lattice_with_quadtree(N0, N, e0, a, n0, T, grid_size)

    # Simulate the exciton hopping and energy relaxation
    exciton_energies = monte_carlo_simulation_with_quadtree(lattice, quadtree, grid_size, n0, a, T, num_excitons)

    # Plot the resulting PL spectrum
    plot_pl_spectrum(exciton_energies)

# Run the simulation
run_simulation_with_quadtree()
```
